```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Ploting a simple 2D line plot
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.plot(x, y)  # plt.plot(x-axis, y-axis)

plt.show()
```

```python
# Data plotting from pandas DataFrame

batsman = pd.read_csv('Data/sharma-kohli.csv')
# batsman.head(5)

plt.plot(batsman['index'],batsman['V Kohli'])
plt.show()
```

```python
# plotting multiple plots

plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.show()
```

```python
# labels title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# colors(hex) and line(width and style) and marker(size)

plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2)

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3,marker='D',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='o')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# legend -> location

plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',label='Rohit')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

# plt.legend(loc='upper right')
plt.legend() # default location is loc='best'
```

```python
# limiting axes
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
plt.ylim(0,75000)   # y-axis limit range
plt.xlim(2017,2019) # x-axis limit range

plt.show()
```

```python
# grid

plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',label='Rohit')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.legend()
plt.grid()
plt.show()
```

```python
# save a figure

plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',label='Rohit')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.legend()
plt.grid()
plt.savefig('sample.png')   # plt.savefig() always runs before plt.show(), other wise it store a blank paper

plt.show()   # plt.show() means ends here
```
