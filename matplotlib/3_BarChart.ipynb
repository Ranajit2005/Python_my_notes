```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# simple bar chart

children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.bar(colors,children)  #plt.bar(x-axis,y-axis)
# plt.bar(colors,children,color=colors)

plt.show()
```

```python
# horizontal bar chart
plt.barh(colors,children,color=colors)
plt.xlabel('Number of children')
plt.ylabel('Colors')
plt.title('Bar chart of children with colors')
plt.show()
```

```python
# bar chart using data

df = pd.read_csv('Data/batsman_season_record.csv')
# np.arange(df.shape[0])
df
```

```python
# np.arange(df.shape[0]) -> convert it in the number and arrange them that we can do operation on it
# df.shape[0] -> number of rows

plt.bar(np.arange(df.shape[0]) - 0.3,df['2015'],width=0.3,color='green',label='2015')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.3,color='red',label='2016')
plt.bar(np.arange(df.shape[0]) + 0.3,df['2017'],width=0.3,color='blue',label='2017')

plt.xticks(np.arange(df.shape[0]), df['batsman'])   # here x-axis values are added by batsman name
plt.xlabel('Batsman')
plt.ylabel('Runs')

plt.legend()
plt.title('Runs scored by batsman in different years')

plt.show()
```

```python
# a problem of overlapping of x-axis values can be solved by rotating the x-axis values
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']

plt.bar(colors,children,color='black')
plt.xticks(rotation='vertical')
plt.show()
```

```python
# Stacked Bar chart

plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016'] + df['2017']),label='2015')

plt.legend()
plt.show()
```
