```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
batters = pd.read_csv('Data/batter.csv')
batters.head()
```

```python
# one figure in subplot

ig, ax = plt.subplots(figsize=(10, 6))     # It returns a tuple containing a figure and axes object(s)

ax.scatter(batters['avg'], batters['strike_rate'], color='red', marker='*')

ax.set_title('Something')
ax.set_xlabel('Avg')
ax.set_ylabel('Strike Rate')

plt.show()
```

```python
# More than one figure in subplot

fig, ax = plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(10,6))   # nrows=2,ncols=1 means 2 rows and 1 column and sharex=True means sharing x axis. nrows,ncols can be accessed by ax[0],ax[1] respectively similar like matrix indexing

# first figure
ax[0].scatter(batters['avg'], batters['strike_rate'], color='red', marker='*')
ax[0].set_title('Something')
ax[0].set_ylabel('Strike Rate')

# second figure
ax[1].scatter(batters['avg'], batters['runs'], color='blue', marker='o')
ax[1].set_title('Something Else')
ax[1].set_xlabel('Avg')
ax[1].set_ylabel('Fours')

plt.show()
```

```python
# Another example of subplot

fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,6))   # we can access by ax[0,0],ax[0,1],ax[1,0],ax[1,1] respectively similar like matrix indexing

# first figure
ax[0,0].scatter(batters['avg'], batters['strike_rate'], color='red', marker='*')
ax[0,0].set_title('Something')
ax[0,0].set_ylabel('Strike Rate')

# second figure
ax[0,1].scatter(batters['avg'], batters['runs'], color='blue', marker='o')
ax[0,1].set_title('Something Else')
ax[0,1].set_ylabel('Runs')

# third figure
ax[1,0].hist(batters['runs'], bins=10, color='green', edgecolor='black')
ax[1,0].set_title('Another Thing')
ax[1,0].set_xlabel('Avg')
ax[1,0].set_ylabel('Fours')

# fourth figure
ax[1,1].hist(batters['avg'], bins=10, color='purple', edgecolor='black')
ax[1,1].set_title('Yet Another Thing')
ax[1,1].set_xlabel('Avg')
ax[1,1].set_ylabel('Sixes')

plt.tight_layout()
plt.show()
```

```python
# Another way of subplot

fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)   # fig.add_subplot(nrows,ncols,index) means words (how many rows, how many columns, which index)
ax1.scatter(batters['avg'],batters['strike_rate'],color='red')

ax2 = fig.add_subplot(2,2,2)
ax2.hist(batters['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batters['avg'])

plt.tight_layout()  # It automatically adjusts subplot params so that the subplot(s) fits in to the figure area.
plt.show()
```
