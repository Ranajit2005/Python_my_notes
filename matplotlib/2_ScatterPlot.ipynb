```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# plt.scatter simple function

x = np.linspace(-10,10,50)  # linspace(start,end,number of points), gives evenly spaced numbers over a specified interval
y = 10*x + 3 + np.random.randint(0,300,50)

plt.scatter(x,y)  # scatter(x-axis,y-axis)
plt.show()

# Scatter plot has same function as 2D plot
```

```python
# plt.scatter on pandas data

df = pd.read_csv('Data/batter.csv')
df.sample(10)

plt.scatter(df['avg'],df['strike_rate'],color='blue',marker='*')
plt.title('Batting Average vs Strike Rate')
plt.xlabel('Batting Average')
plt.ylabel('Strike Rate')
plt.grid()

plt.show()
```

```python
# slower
plt.scatter(df['avg'],df['strike_rate'],s=df['runs']*0.09)  # s is for size of marker
```

```python
# scatterplot using plt.plot

# faster

plt.plot(df['avg'],df['strike_rate'],'o')
plt.show()
```
