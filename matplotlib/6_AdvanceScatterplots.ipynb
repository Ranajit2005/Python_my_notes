```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
iris = pd.read_csv('Data/iris.csv')
iris.sample(5)
```

```python

iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
# colur only use on number values, so replace dofferent category by numbers
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.7)  #c is for color, cmap is for color map, alpha is for transparency
plt.title('Sepal Length vs Petal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar() # show color scale
plt.show()
```

```python
# figure size

plt.figure(figsize=(15,7))   # figsize=(width,height)

plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
plt.show()
```

```python
# Scatter plot with point value name and size

x = [1,2,3,4]
y = [5,6,7,8]

plt.scatter(x,y)
plt.text(1,5,'Point 1')
plt.text(2,6,'Point 2')
plt.text(3,7,'Point 3')
plt.text(4,8,'Point 4',fontdict={'size':12,'color':'brown'})
plt.show()
```

```python
# Working with dataset
batters = pd.read_csv('Data/batter.csv')
sample_df = batters.head(100).sample(25,random_state=5)   # random_state is for reproducibility, and sample(how many rows you want to select, random_state=any number)
sample_df
```

```python
# Scatter plot with point value name and size using dataset

plt.figure(figsize=(19,10))

plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs']) # s means size of the points

for i in range (sample_df.shape[0]):   # here every batsman has their differants values, means differents name, so we use for loop and every time pass the .values[i]
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])

plt.xlabel('Average')
plt.ylabel('Strike Rate')
plt.title('Batter Analysis')

plt.show()
```

```python
# Horizontal and Vertical lines

plt.figure(figsize=(18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

# h for horizontal line, v for vertical line
plt.axhline(130,color='red')
plt.axhline(140,color='green')
plt.axvline(30,color='red')

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```
