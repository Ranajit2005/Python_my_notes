```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
batters = pd.read_csv('Data/batter.csv')
batters.head()
```

### 3D Scatter Plot

```python
# fig = plt.figure()
fig = plt.figure(figsize=(10,7))

ax = plt.subplot(projection='3d')

ax.scatter3D(batters['avg'],batters['strike_rate'],batters['runs'],color='blue')

ax.set_title('3D Scatter Plot of Batting Average, Strike Rate and Runs')
ax.set_xlabel('Batting Average')
ax.set_ylabel('Strike Rate')
ax.set_zlabel('Runs')

plt.show()
```

### 3D Line Plot

```python
x = [0,1,5,25]
y = [0,10,13,0]
z = [0,13,20,9]

fig = plt.figure(figsize=(10,7))

ax = plt.subplot(projection='3d')

# ax.scatter3D(x,y,z,color='red',s=100)
ax.plot3D(x,y,z,color='green')

ax.set_title('3D Line Plot')
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')

plt.show()
```

### 3D Surface Plots

```python
# Create an example data

x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

X,Y = np.meshgrid(x,y)  # Create a grid of x and y values

Z = np.sin(np.sqrt(X**2 + Y**2))

# Plotting the surface
fig = plt.figure(figsize=(10,7))

ax = plt.subplot(projection='3d')

ax.plot_surface(X,Y,Z, cmap='viridis')

ax.set_title('3D Surface Plot')
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')

plt.show()
```

```python
z = np.sin(X) + np.cos(Y)

fig = plt.figure(figsize=(10,7))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(X,Y,z,cmap='viridis')
fig.colorbar(p)  # Add a color bar which maps values to colors.

plt.show()
```

```python
z = np.sin(X) + np.tan(X)

fig = plt.figure(figsize=(10,7))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(X,Y,z,cmap='viridis')
fig.colorbar(p)

plt.show()
```

### Contour Plots

```python
fig = plt.figure(figsize=(10,7))

ax = plt.subplot()

p = ax.contour(X,Y,z,cmap='viridis')  # Contour plot
fig.colorbar(p)
plt.show()
```

```python
# z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contourf(X,Y,z,cmap='viridis')   # Filled contour plot
fig.colorbar(p)
plt.show()
```
