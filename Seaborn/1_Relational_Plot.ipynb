```python
tips = sns.load_dataset('Data/tips')
tips
```

```python
# scatter plot -> axes level function
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
# hue is used for categorical variable, size is used for numerical variable, style is used for categorical variable

sns.scatterplot(data=tips,x='total_bill',y='tip',hue='smoker',size='size',sizes=(20,200))
```

```python
# hue is used for categorical variable, size is used for numerical variable, style is used for categorical variable

sns.scatterplot(data=tips,x='total_bill',y='tip',hue='smoker',style='time',size='size',sizes=(20,200))
```

```python
# relplot -> figure level -> square shape
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='smoker',style='time',size='size',sizes=(20,200))
# here the function name is relplot and kind is scatter
```

```python
# line plot
gap = px.data.gapminder()
temp_df = gap[gap['country'] == 'India']
temp_df.head()
```

```python
# axes level function
sns.lineplot(data=temp_df, x='year', y='lifeExp')
```

```python
# using relplot, figure level function
sns.relplot(data=temp_df, x='year', y='lifeExp', kind='line')
```

```python
# hue -> style
temp_df = gap[gap['country'].isin(['India','Brazil','Germany'])]
temp_df.head()
```

```python
sns.relplot(data=temp_df, kind='line', x='year', y='lifeExp', hue='country', style='continent',size='continent')
```

```python
sns.lineplot(data=temp_df, x='year', y='lifeExp', hue='country')
```

```python
# facet plot -> figure level function -> work with relplot
# it will not work with scatterplot and lineplot
sns.relplot(data=tips, x='total_bill', y='tip', kind='line', col='sex', row='day')
# col -> column, row -> row, and it will create multiple plots based on the unique values of the column and row
```

```python
# col wrap -> it will wrap the columns into multiple rows after the specified number
sns.relplot(data=gap, x='lifeExp', y='gdpPercap', kind='scatter', col='year', col_wrap=3)
```
