```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
import plotly.express as px
```

### Two types 
    -> Heatmap
    -> Clustermap

```python
gap = px.data.gapminder()
gap.head()
```

```python
# Heatmap

# Plot rectangular data as a color-encoded matrix
temp_df = gap.pivot(index='country',columns='year',values='lifeExp')    # index shows y axis, columns shows x axis, values shows the value to be filled in the matrix

# axes level function
plt.figure(figsize=(15,15))
sns.heatmap(temp_df)
```

```python
# annot
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country',columns='year',values='lifeExp')

plt.figure(figsize=(15,15))
sns.heatmap(temp_df,annot=True,linewidth=0.5, cmap='summer')  # annot -> write the data value in each cell, linewidth -> width of lines that will divide each cell, cmap -> colormap to use
```

```python
# Clustermap

# Plot a matrix dataset as a hierarchically-clustered heatmap.

# This function requires scipy to be available.

iris = px.data.iris()
iris.head()
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]])
```
