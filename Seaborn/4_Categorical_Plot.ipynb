```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
# import datasets
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
```

```python
sns.scatterplot(data=tips, x='total_bill',y='tip')
```

```python
# strip plot
# axes level function
sns.stripplot(data=tips,x='day',y='total_bill') # Seaborn automatically adds jitter, so points don't overlap
```

```python
# using catplot
# figure level function
sns.catplot(data=tips, x='day',y='total_bill',kind='strip')
```

```python
# jitter
sns.catplot(data=tips, x='day',y='total_bill',kind='strip',jitter=0.15,hue='sex')
```

```python
# swarmplot -> better than stripplot because it avoids overlapping points and is more informative, but it is slower
# figure level function
sns.catplot(data=tips, x='day',y='total_bill',kind='swarm')
```

```python
# axes level function
sns.swarmplot(data=tips, x='day',y='total_bill')
```

```python
# hue
sns.swarmplot(data=tips, x='day',y='total_bill',hue='sex')
```
